```python
import sys
from pathlib import Path
from datetime import datetime
import polars as pl
import matplotlib.pyplot as plt
from glob import glob
import os

RAIZ = Path.cwd()
if RAIZ.name == "notebooks":
    RAIZ = RAIZ.parent
```

```python
rutas = glob(r"C:\Users\josearey\Documents\proyecto datascience - modelo de fuga\raw\wong_dia_*.parquet")
esquemas = []
for ruta in rutas:
    df = pl.scan_parquet(ruta)
    esquemas.append(df.collect_schema)
    r = df.select(
        pl.len().alias("filas"),
        pl.col('cod_dia').min().alias('fecha_minima'),
        pl.col('cod_dia').max().alias('fehca_maxima'),
    ).collect()
    print(f"{os.path.basename(ruta):<24} {r['filas'][0]:>12,}  {r['fecha_minima'][0]} -> {r['fehca_maxima'][0]}")

print("esquemas identicos:", all(e == esquemas[0] for e in esquemas))
```
